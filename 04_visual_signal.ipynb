```python
# Input file from keyword detection step
input_file = "/content/drive/MyDrive/Dark Pattern Thesis/Batch 2/batch2_keyword_outputs/batch2_keyword_detection_results.xlsx"

# Output folder for visual signal analysis
output_folder = "/content/drive/MyDrive/Dark Pattern Thesis/Batch 2/batch2_visual_signal_outputs"

# Create output folder if it does not exist
os.makedirs(output_folder, exist_ok=True)

print("Input file exists:", os.path.exists(input_file))
print("Output folder:", output_folder)
```

```python
df = pd.read_excel(input_file)

print("Rows loaded:", len(df))
df.head()
```

```python
def contains_any(text, keywords):
    """
    Checks whether any keyword appears in the text.
    Returns Yes or No.
    """

    if pd.isna(text):
        text = ""

    text = str(text).lower()

    for keyword in keywords:
        if keyword.lower() in text:
            return "Yes"

    return "No"
```

```python
# Combine OCR text and screenshot filename into one searchable text field
df["Combined_Text"] = (
    df["Screenshot_ID"].astype(str).str.lower()
    + " "
    + df["OCR_Text"].astype(str).str.lower()
)
```

```python
# Candidate fake urgency visual/textual signals
df["Countdown Timer"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "countdown", "timer", "00:", "01:", "02:", "03:", "ends in"
    ])
)

df["Low Stock Text"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "low stock", "only", "left", "few left", "almost gone",
        "selling fast", "stock"
    ])
)

df["Limited Time Badge"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "limited time", "limited", "today only", "deal ends",
        "ends soon", "last chance", "early bird"
    ])
)

df["Urgency Emphasis"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "hurry", "limited", "only", "left", "ends", "last chance",
        "selling fast", "order within", "deal ends", "early bird"
    ])
)

# Candidate visual misdirection signals
df["Discount Banner"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "discount", "sale", "deal", "offer", "save", "clearance",
        "% off", "off"
    ])
)

df["Crossed-Out Price"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "crossed", "crossed-out", "was", "was price", "old price",
        "original price", "list price", "before"
    ])
)

df["Price Anchor"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "was", "now", "original price", "list price", "save",
        "reduced", "before"
    ])
)

df["Dominant Buy Button"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "buy now", "add to cart", "add to bag", "shop now",
        "checkout", "order now", "continue"
    ])
)

df["Sale Colour Emphasis"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "sale", "deal", "discount", "off", "%", "clearance",
        "black friday", "flash sale"
    ])
)

# Candidate obstruction signals
df["Popup Present"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "popup", "pop up", "modal", "welcome", "newsletter",
        "subscribe", "sign up", "signup", "register"
    ])
)

df["Popup Blocks Content"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "popup", "modal", "blocks content", "blocking",
        "newsletter", "welcome", "sign up", "subscribe"
    ])
)

df["Forced Signup"] = df["Combined_Text"].apply(
    lambda x: contains_any(x, [
        "sign up", "signup", "register", "create account",
        "subscribe", "newsletter", "join now"
    ])
)

df.head()
```

```python
fake_urgency_visual_cols = [
    "Countdown Timer",
    "Low Stock Text",
    "Limited Time Badge",
    "Urgency Emphasis"
]

misdirection_visual_cols = [
    "Discount Banner",
    "Crossed-Out Price",
    "Price Anchor",
    "Dominant Buy Button",
    "Sale Colour Emphasis"
]

obstruction_visual_cols = [
    "Popup Present",
    "Popup Blocks Content",
    "Forced Signup"
]

df["Visual Fake Urgency Detected"] = df[fake_urgency_visual_cols].apply(
    lambda row: "Yes" if "Yes" in row.values else "No",
    axis=1
)

df["Visual Misdirection Detected"] = df[misdirection_visual_cols].apply(
    lambda row: "Yes" if "Yes" in row.values else "No",
    axis=1
)

df["Visual Obstruction Detected"] = df[obstruction_visual_cols].apply(
    lambda row: "Yes" if "Yes" in row.values else "No",
    axis=1
)

df.head()
```

```python
visual_columns = (
    fake_urgency_visual_cols
    + misdirection_visual_cols
    + obstruction_visual_cols
)

summary_rows = []

for col in visual_columns:
    count_yes = (df[col] == "Yes").sum()
    percentage = round(count_yes / len(df) * 100, 1)

    summary_rows.append({
        "Visual Signal": col,
        "Detected Count": count_yes,
        "Percentage": percentage
    })

visual_summary = pd.DataFrame(summary_rows)

visual_summary
```

```python
output_file = os.path.join(
    output_folder,
    "batch2_visual_signal_analysis_results.xlsx"
)

summary_file = os.path.join(
    output_folder,
    "batch2_visual_signal_analysis_summary.xlsx"
)

df.to_excel(output_file, index=False)
visual_summary.to_excel(summary_file, index=False)

print("Saved visual signal results:")
print(output_file)

print("Saved visual signal summary:")
print(summary_file)
```
