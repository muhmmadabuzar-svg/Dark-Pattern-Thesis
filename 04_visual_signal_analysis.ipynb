```python
import pandas as pd

input_file = "/content/drive/MyDrive/Dark Pattern Thesis/ocr_outputs/first_batch_keyword_detection_results.xlsx"

output_file = "/content/drive/MyDrive/Dark Pattern Thesis/ocr_outputs/first_batch_visual_signal_coding.xlsx"

df = pd.read_excel(input_file)

print("Total rows:", len(df))
df.head()
```

```python
visual_columns = [
    # Fake urgency visual/textual visibility
    "countdown_timer_visible",
    "low_stock_text_visible",
    "limited_time_badge_visible",
    "urgency_text_visually_emphasized",

    # Visual misdirection
    "large_discount_banner",
    "large_sale_percentage",
    "crossed_out_price",
    "price_anchor_visible",
    "dominant_buy_button",
    "sale_color_emphasis",

    # Obstruction
    "popup_present",
    "popup_blocks_content",
    "dimmed_background",
    "small_close_button",
    "forced_signup_or_register",
    "app_download_prompt",

    # General coding notes
    "visual_coding_notes"
]

for col in visual_columns:
    if col not in df.columns:
        df[col] = ""

df.head()
```

```python
df.to_excel(output_file, index=False)

print("Visual signal coding file created:")
print(output_file)
```
